#### Step 2: 
##### Handling Missing Values

Missing values can introduce inaccuracies and create bias in the analysis. So, it is important identify them and handle them appropriately. To do this, I will get the sum of all isnull rows for each column in the dataframely

```python
# checking for missing values
df_copy.isnull().sum()
```

Observation:

There are 9 columns with missing/null values, majority of which have less than 4 missing values. I will be using the interpolation method to replace the missing values. Interpolation will estimate the missing values based on surrounding data points, which can perserve the data's natural trend patterns. By using nearby values, interpolation will also maintain some of the variability in the data, avoiding the reduction in the variability that can occur with the mean/median imputation. 

However, for the one variable - 'authorities_contracted' - which has 93 missing values, more investigation is required in order to make an informed decision as to how to handle this column. 

```python
# interpolation of the missing/null values in all the other columns, excluding 'authorities_contacted'
df_copy['age'] = df_copy['age'].interpolate(method='linear')
df_copy['policy_deductable'] = df_copy['policy_deductable'].interpolate(method='linear')
df_copy['policy_annual_premium'] = df_copy['policy_annual_premium'].interpolate(method='linear')
df_copy['insured_education_level'] = df_copy['insured_education_level'].interpolate(method='linear')
df_copy['insured_hobbies'] = df_copy['insured_hobbies'].interpolate(method='linear')
df_copy['capital-gains'] = df_copy['capital-gains'].interpolate(method='linear')
df_copy['incident_state'] = df_copy['incident_state'].interpolate(method='linear')
df_copy['total_claim_amount'] = df_copy['total_claim_amount'].interpolate(method='linear')
df_copy['injury_claim'] = df_copy['injury_claim'].interpolate(method='linear')
df_copy['property_claim'] = df_copy['property_claim'].interpolate(method='linear')
```

```python
# verifying that the missing data has been filled
df_copy.isnull().sum()
```

Observation:

We note that 'insured_education_level' , 'insured_hobbies' & 'incident_state' did not fill in missing values during the interpolation approach. This is because interpolation is best suited for numerical values and these variables seem to be categorical in nature. So, for these I will use the mode imputation which will replace the missing values with the most frequently occuring value in the column. This is a common approach for categorical data, and given that at most for one column I will only be replacing 2 missing values, I feel comfortable to do so.

```python
# Replace missing values with the mode for categorical columns
df_copy['insured_education_level'].fillna(df_copy['insured_education_level'].mode()[0], inplace=True)
df_copy['insured_hobbies'].fillna(df_copy['insured_hobbies'].mode()[0], inplace=True)
df_copy['incident_state'].fillna(df_copy['incident_state'].mode()[0], inplace=True)
```

```python
# verifying that the missing data has been filled
df_copy.isnull().sum()
```

Observation:

I have successfully replaced the missing values for this dataset, with the exception of 'authorities_contracted'. Given that this column has 93 values missing, which is about 9.3% of the dataset, I don't feel comfortable imputing these values. It is therefore important to assess the importance of this variable on our investigation for detecting fraudulent claims. If the column is not at all important, I will feel most comfortable dropping the column all together.

#### Step 4:
##### Handling Outliers

Handling outliers is a critical step in ensuring that your data and models are robust, accurate, and reliable.

I will be using the IQR method to identify any outliers in the data. 

The IQR method is a robust statisitcal technique for identifying outliers by leveraging from the spread of the data. It is useful for identifying extreme values that deviate from the majority of the dataset, helping me clean the data for accurate analysis and modeling.


```python
# Identify and handle outliers in all the numeric columns
numeric_cols = ['months_as_customer', 'age', 'policy_deductable', 'policy_annual_premium', 'umbrella_limit',
    'insured_zip', 'capital-gains', 'capital-loss', 'incident_hour_of_the_day', 'number_of_vehicles_involved',
    'bodily_injuries', 'witnesses', 'total_claim_amount', 'injury_claim', 'property_claim', 'vehicle_claim',
    'auto_year']

# Calculate quartiles and IQR for each numeric column
for col in numeric_cols:
    Q1 = df_copy[col].quantile(0.25)
    Q3 = df_copy[col].quantile(0.75)
    IQR = Q3 - Q1
    
    # Identify outliers
    outliers = (df_copy[col] < Q1 - 1.5 * IQR) | (df_copy[col] > Q3 + 1.5 * IQR)
    
    # Remove outliers
    df_copy = df_copy[~outliers]

df_copy.shape
```

Observation:

This method has allowed me to remove outliers. In total, 21.8% of the dataset has been removed at this stage. I am confident in moving forward with only 782 observations as this is still over 70% of the dataset. Additionally, this ensure that my data is more aligned for robust analysis and accurate modeling.
